## **Sorting Algorithms**

This section will explain some of the popular sorting algorithms, providing their steps in pseudocode, a simple example, and a Python implementation which allow you to practice these algorithms yourself.

Before we start, we should explain and define a few terms.

* **Algorithm:** An algorithm is a finite, deterministic, and effective problem-solving method suitable for implementation as a computer program. We can define an algorithm by describing a procedure for solving a problem in a
natural language, flowdiagram, pseducode, or just by writing a computer program that implements the procedure. An example is the Euclid’s algorithm for finding the greatest common divisor of two numbers, a variant of which was devised
over 2,300 years ago.  (From Algorithms by Sedgewick)


```python
def gcd(p, q):
    if q == 0:
        return p
    r = p % q
    return gcd(q, r)

# Example usage:
p = 48
q = 18
result = gcd (p, q)
print(f"The GCD of {p} and {q} is: {result}")
```

* **Sorting:** Sorting is the process of arranging a collection of data into a specific, meaningful order. This is typically done to make the data easier to search, analyze, or visualize.

   - It is a fundamental building block of computing. For example, a search engine can find information faster if its index is sorted, and your banking app sorts transactions so you can see your most recent spending at a glance.

   - Most sorting is either numerical (1, 2, 3...) or lexicographical (A, B, C...). Data can be sorted in ascending order (lowest to highest) or descending order (highest to lowest). Efficiency: In computer science, different algorithms (like QuickSort, MergeSort, or Bubble Sort) are used depending on the size and type of data to optimize speed and memory usage.
   - Common Examples:
     * Numerical: Arranging a list of prices from cheapest to most expensive (\$12, \$45, \$102).
     * Alphabetical: Organizing a contact list by last name (Adams, Brown, Clark).
     * Chronological: Sorting emails or files by the date they were created or received.


* **Algorithm analysis**: the process of evaluating the efficiency of an algorithm, focusing on its resource requirements e.g. time (speed) and space (memory) when the input size increases. This analysis is hardware independent.  
    - **Big O Notation**: This mathematical shorthand describes the "worst-case scenario" or the upper bound allowing developers to predict how an algorithm will perform under heavy loads without needing to run it on every possible hardware configuration.
    - Examples:
      - **$O(1)$ Constant:** Accessing a specific element in an array by its index. It takes the same time regardless of array size.
      - **$O(\log(n))$ Logarithmic:** Finding a word in a dictionary using Binary Search (splitting the search area in half repeatedly).
      - **$O(n)$ Linear:** Searching for a specific name in an unsorted list by checking every entry from start to finish.
      - **$O(n^2)$ Quadratic:** Comparing every item in a list to every other item, such as in a simple Bubble Sort.
      - **$O(2^n)$ Exponential:** The execution time doubles with every single addition to the input data set ($n$). These algorithms become impractical for even relatively small values of $n$. For example cracking a Password, if you try to guess a 10-digit password by testing every possible combination, adding just one more digit to the password it makes it much much harder, see the code example.
    - Similarly:     
      - **Big Omega $\Omega$**: Describes the "best-case scenario" or the lower bound.
      - **Big Theta ($\Theta$):** Describes the "average-case scenario" or the tight bound. If an algorithm's best case and worst case grow at the same rate, we use Theta to say "it will always perform exactly like this.
      Example: Printing every element in an array. There is no "lucky" way to do this faster—you must touch every item once. Therefore, it is $\Theta(n)$ because the best, worst, and average cases are all linear.      


      

```python
# Example of algorithms analysis
# Notice how the number change for each class when we increase n
import math

def get_growth_value(n, complexity_type):
    try:
        if complexity_type == "Constant":    # O(1)
            return 1
        elif complexity_type == "Logarithmic": # O(log n)
            return round(math.log2(n), 2) if n > 0 else 0
        elif complexity_type == "Linear":      # O(n)
            return n
        elif complexity_type == "Quadratic":   # O(n^2)
            return n ** 2
        elif complexity_type == "Exponential": # O(2^n)
            # Prevent the program from crashing on huge numbers
            if n > 100:
                return "Value too large to compute (Infinity)"
            return 2 ** n
    except Exception as e:
        return str(e)

# Testing across different scales
test_values = [10, 20, 100, 1000]
complexities = ["Constant", "Logarithmic", "Linear", "Quadratic", "Exponential"]

for n in test_values:
    print(f"--- Results for n = {n} ---")
    for comp in complexities:
        result = get_growth_value(n, comp)
        print(f"{comp:12}: {result}")
    print()




```

### **(1) Selection Sort**

One of the simplest sorting algorithms that works by repeatedly selecting the smallest remaining item.

#### **How it works**

- First, find the smallest item in the array and exchange it with the first entry (itself if the first entryis already the smallest).
- Then, find the next smallest item and exchange it with the second entry.
- Continue in this way until the entire array is sorted.

#### **Example:**

Assuming we have a list [64, 25, 12, 22, 11].
- First we compare 64 with all remaining element 25 to 11 and save the index of the minimum element, let's call it $idx_min$ (assuming the first index is 0).   
  - At the first comprison 25 is smaller than 25 so our $idx_min$ is changed from 0 to 1.
  - At the second comprison 12 is smaller than 25 so our $idx_min$ is changed from 1 to 2.
  - At the third comprison 22 is greater than 12 so our $idx_min$ remains the same
  - At the forth comprison 11 is smaller than 12 so our $idx_min$ is changed from 2 to 4
  - Since we reached the final element we now swap the first element with idx_min element i.e. 64 and 11. Our updated list will look like this after the first round:
     [11, 25, 12, 22, 64]
- Now we repeat the same process but we start from the second element i.e. 25 and compare with remaining elements 12 to 64

The python code section shows all the details.

#### **Pseudocode:**

Algorithm SelectionSort:

$$
\begin{array}{l}
   \hline  \mathbf{Algorithm:} \text{ SelectionSort}(L, n) \\   \hline
       \textbf{for } i \leftarrow 0 \text{ to } n - 2 \textbf{ do:} \\
       \quad \textit{min_idx} \leftarrow i \\
       \quad \textbf{for } j \leftarrow i + 1 \text{ to } n - 1 \textbf{ do:} \\
       \quad \quad \textbf{if } L[j] < L[\textit{min_idx}] \textbf{ then} \\
             \quad \quad \quad \textit{min_idx} \leftarrow j \\
       \quad \text{swap}(L[i], L[\textit{min_idx}]) \\
    \hline
\end{array}
$$


#### **Algorithm analysis:**

Selection Sort performance is generally **inefficient** on large lists compared to algorithms like QuickSort or MergeSort.

- Comparisons:
    - The number of comparisons is independent of the initial order of the input.
   - For an array of size $n$, the total number of comparisons is:
$\sum_{i=0}^{n-2} (n-1-i) \approx \frac{n(n-1)}{2} \approx \frac{n^2}{2}$
- Exchanges:
  - There are exactly $n-1$ exchanges (swaps), which is $O(n)$. This makes Selection Sort useful when memory write operations are significantly more expensive than reads.
- Complexity Classes:
  - Worst-case Time $O(n^2)$: Occurs regardless of data order due to nested loops $(n-1)(n-2)$.
  - Average-case Time $\Theta(n^2)$: The inner loop always runs $(n-1)(n-2)$ times.
  - Best-case Time $\Omega(n^2)$: Even if the array is sorted, the algorithm still scans for the minimum using $(n-1)(n-2)$ times.
- Space Complexity$O(1)$: It is an in-place algorithm, requiring only a constant amount of extra space.

#### **Pyhton code**

```python
# Selection Sort example
# TODO: add counters to count the number of comparisons and the number of exchange
import random
def selection_sort(L, n, logging=1):
    for i in range(n-1): # 0 to n-1
        # Assume the current position is the minimum
        min_idx = i

        # Scan the rest of the array to find/select the actual minimum
        for j in range(i + 1, n): # i+1 to n-1
            # compare with current minimum
            if logging:
                print("i,j ", i, j, " min_idx before: ", min_idx)

            if L[j] < L[min_idx]:
                min_idx = j

            if logging:
                print("i,j ", i, j, " min_idx  after: ", min_idx)

        # Swap the found minimum element with the first element of the unsorted part
        if logging:
            print("L before: ", L)
            print("Swapping: ", L[i], L[min_idx])

        L[i], L[min_idx] = L[min_idx], L[i]

        if logging:
            print("L  after: ", L)

    if logging:
        print("Selection sorting is completed!")

    return L

# Example Usage:
unsorted_data = [64, 25, 12, 22, 11]
# # we can also generate random list between 1 and 100 with n length
# n= 10
# unsorted_data = [random.randint(1, 100) for _ in range(n)]
n = len(unsorted_data)
sorted_data = selection_sort(unsorted_data.copy(), n)
print("-----------------------------")
print(f"unsorted_data: {unsorted_data}")
print(f"Sorted array : {sorted_data}")
```

### **(2) Insertion Sort**

The algorithm that people often use to sort cards by consider the cards one at a time, inserting each into its proper place among those already
considered (keeping them sorted).

The insertion sort is much faster if the most entries are ordered or in reverse order.

#### **How it works:**

- Start with the second element (index 1) and assume the first element is a "sorted" sub-list of length 1.
- Compare the current element (the "key") with the elements in the sorted sub-list to its left.
- If the element in the sorted sub-list is larger than the key, shift that element one position to the right.
- Continue shifting until you find the correct spot for the key, then insert it.
- Repeat this process for all elements until the entire array is sorted.

#### **Example:**

Assuming we have a list: [64, 25, 12, 22, 11]

- First Pass: The key is 25. Compare with 64. Since 64 > 25, move 64 to the right and insert 25. List: [25, 64, 12, 22, 11]
- Second Pass: The key is 12. Compare with 64 (shift) and 25 (shift). Insert 12 at the beginning. List: [12, 25, 64, 22, 11]
- Third Pass: The key is 22. Compare with 64 (shift) and 25 (shift). Since 12 is smaller than 22, stop shifting and insert 22. List: [12, 22, 25, 64, 11]
- Fourth Pass: The key is 11. Compare with 64, 25, 22, and 12. Shift all and insert 11 at the start. Sorted List: [11, 12, 22, 25, 64]

The python code section shows all the details.

#### **Pseudocode:**

Algorithm InsertionSort:


$$
\begin{array}{l}
   \hline    \mathbf{Algorithm:} \text{ InsertionSort}(L, n) \\ \hline
   \textbf{for } i \leftarrow 1 \text{ to } n-1 \textbf{ do:} \\
          \quad \text{key} \leftarrow L[i] \\
          \quad j \leftarrow i - 1 \\
          \quad \textbf{while } j \ge 0 \text{ and } L[j] > \text{key} \textbf{ do:} \\
            \quad \quad L[j+1] \leftarrow L[j] \\
            \quad \quad j \leftarrow j - 1 \\
          \quad L[j+1] \leftarrow \text{key} \\
    \hline
\end{array}
$$



#### **Algorithm analysis:**

Insertion Sort is highly efficient for small data sets or lists that are already nearly sorted.

- Comparisons:
  -In the worst case (reverse order), the number of comparisons  for an array of size $n$ is: $\sum_{i=1}^{n-1} (i) \approx \frac{n(n-1)}{2} \approx \frac{n^2}{2}$. In a randomly ordered array, the element is equally likely to land in any of those $i+1$ spots. To find the average number of shifts for that specific step, we sum the possible shifts and divide by the number of possibilities:$$\text{Average shifts for element } i = \frac{0 + 1 + 2 + \dots + i}{i + 1}$$
  Using the arithmetic series formula $\sum_{k=1}^n k = \frac{n(n+1)}{2}$, the numerator becomes $\frac{i(i+1)}{2}$. When we divide that by $(i+1)$, the $(i+1)$ terms cancel out:$$\frac{\frac{i(i+1)}{2}}{i+1} = \frac{i}{2}$$. We sum these individual averages for every element from the first to the last ($n$):$$\sum_{i=1}^{n-1} \frac{i}{2} = \frac{1}{2} \sum_{i=1}^{n-1} i = \frac{1}{2} \left( \frac{(n-1)n}{2} \right) = \frac{n^2 - n}{4} \approx \frac{n^2}{4}$$

- Exchanges:
  - It requires $n^2/4$ exchanges to sort a randomly ordered array of length n with distinct keys, on the average. The worst case is $n^2/2$ exchanges and the best case is and 0 exchanges.
- Complexity Classes:
  - Worst-case Time $O(n^2)$: Occurs when the array is sorted in reverse order $n^2/4$.
  - Average-case Time $\Theta(n^2)$: On average, we shift half of the elements in the sorted sub-list $n^2/4$.
  - Best-case Time $\Omega(n)$: Occurs when the array is already sorted. The inner loop never executes its body, only performing one comparison per element $(n-1)$.  

- Space Complexity$O(1)$: It is an in-place algorithm, requiring only a constant amount of extra space.




#### **Python code**

```python
# Insertion Sort example
# TODO: add counters to count the number of comparisons and the number of exchange
import random
def insertion_sort(L,n, logging=1):
    for i in range(1,n): # 1 to n
        # Assume the current position is the key
        key = L[i]
        # Move elements of L[0..i-1] that are greater than key
        # to one position ahead of their current position
        j = i - 1
        # Scan the left side of the array to insert the current value
        if logging:
           print("i,j ", i,j," L before: ",L)
        while j>=0 and key <L[j]: # j=i-1 to 0
            # shift and move to the next element in the left
            L[j + 1] = L[j]
            j -= 1
            if logging:
               print("i,j ", i,j," L after:  ",L)
        # Insert the key to the final location
        if logging:
           print("Final insertion: ",L[j+1], key)
           print("L before: ",L)
        L[j + 1] = key
        if logging:
           print("L  after: ",L)
    print("Insertion sorting is completed!")
    return L

# Example Usage:
unsorted_data = [64, 25, 12, 22, 11]
# # we can also generate random list between 1 and 100 with n length
# n= 10
# unsorted_data = [random.randint(1, 100) for _ in range(n)]
n = len(unsorted_data)
sorted_data = insertion_sort(unsorted_data.copy(), n)
print("-----------------------------")
print(f"unsorted_data: {unsorted_data}")
print(f"Sorted array : {sorted_data}")
```

### **(3) Shell Sort**

Shellsort is a simple extension of insertion sort that gains speed by allowing exchanges of array entries that are far apart, to produce partially sorted arrays that can be efficiently sorted, eventually by insertion sort. The idea is to rearrange the array to give it the property that taking every hth entry
(starting anywhere) yields a sorted subsequence. Such an array is said to be h-sorted.

#### **How it works**
Shell Sort is essentially an optimized version of Insertion Sort. It improves efficiency by allowing the exchange of items that are far apart.
- Choose an initial gap size (often starting at $n/2$).
- Divide the list into several sub-lists where elements in each sub-list are separated by the gap distance.
- Sort these sub-lists using Insertion Sort.
- Reduce the gap (usually by half) and repeat the process.
- The final pass uses a gap of 1, which is a standard Insertion Sort, but the list is already nearly sorted by then, making it very fast.

#### **Example**

Assuming we have a list: [64, 25, 12, 22, 11] and $n=5$. Initial gap $h = \lfloor 5/2 \rfloor = 2$.
- First Pass (Gap = 2):
  - Sub-list 1: indices (0, 2, 4) -> [64, 12, 11]. Sorts to [11, 12, 64].
  - Sub-list 2: indices (1, 3) -> [25, 22]. Sorts to [22, 25].
  - List becomes: [11, 22, 12, 25, 64]
- Second Pass (Gap = 1):
  - Perform a standard Insertion Sort on the entire list.
  - Because the list is already nearly sorted ([11, 22, 12, 25, 64]), very few shifts are required.
  - Final Sorted List: [11, 12, 22, 25, 64]

#### **Pseudocode**
Algorithm ShellSort:

$$\begin{array}{l}
  \hline \mathbf{Algorithm:} \text{ ShellSort}(L, n, h) \\  \hline
     \textbf{while } h < n // 3 \textbf{ do} \\
           \quad h \leftarrow 3 * h + 1 \\
     \textbf{while } h \ge 1 \textbf{ do} \\
          \quad \textbf{for } i \leftarrow h \text{ to } n - 1 \textbf{ do} \\
                  \quad \quad j \leftarrow i \\
                  \quad \quad \textbf{while } j \ge h \text{ and } L[j] < L[j - h] \textbf{ do} \\
                       \quad \quad \quad \text{swap } L[j], L[j - h] \\
                       \quad \quad \quad j \leftarrow j - h \\
          \quad h \leftarrow h // 3 \\
\hline
\end{array}$$

#### **Algorithm analysis:**

Shell Sort’s performance is highly dependent on the gap sequence used.

- Comparisons and Exchanges:Unlike Insertion Sort, Shell Sort moves elements long distances to their destination in the early stages. This drastically reduces the total number of inversions that the final $h=1$ pass has to handle.
- For the Shell's sequence ($n/2, n/4, \dots, 1$), the number of operations is significantly lower than $n^2$ for average cases.
- Complexity Classes:
  - Worst-case Time: $O(n^2)$ for the standard $n/2$ sequence. However, with better sequences like Sedgewick's, it can reach $O(n^{4/3})$ or better.
  - Average-case Time: $\Theta(n^{3/2})$ or $\Theta(n \log^2 n)$ depending on the gap sequence.
  - Best-case Time: $\Omega(n \log n)$. This occurs if the array is already sorted, as it still performs comparisons across the gap passes.
  - Space Complexity $O(1)$: Like Insertion Sort, it is an in-place sorting algorithm and requires no additional memory relative to the size of the input.

#### **Python code**

```python
# SHELL Sort example
# TODO: add counters to count the number of comparisons and the number of exchange
import random, time
def shell_sort(L,n,h, logging=1):
    # Generate the initial gap value,Knuth sequence: 1, 4, 13, 40, ...
    while h < n // 3:
            h = 3 * h + 1
    print("h:", h)
    while h >= 1:
        # h-sort the array
        for i in range(h, n):
            # Insert L[i] among L[i-h], L[i-2*h], L[i-3*h]...
            j = i
            if logging:
               print("i,j,h ", i,j,h," L before: ",L)
            while j >= h and L[j] < L[j - h]:
                    # Pythonic swap (the exch function in the Java code)
                    L[j], L[j - h] = L[j - h], L[j]
                    j -= h
                    if logging:
                       print("i,j,h ", i,j,h," L after:  ",L)
        # Move to the next smaller gap
        h //= 3
    print("Shell sorting is completed!")
    return L

# Example Usage:
unsorted_data = [64, 25, 12, 22, 11]
# # we can also generate random list between 1 and 100 with n length
# n= 10
# unsorted_data = [random.randint(1, 100) for _ in range(n)]
n = len(unsorted_data)
h = n//2
sorted_data = shell_sort(unsorted_data.copy(), n, h)
print("-----------------------------")
print(f"unsorted_data: {unsorted_data}")
print(f"Sorted array : {sorted_data}")
```

#### **Insertion vs Shell sorting**

```python
# Comparing Insertion vs Shell sort using larger list:
n= 100
unsorted_data = [random.randint(1, 1000) for _ in range(n)]

n = len(unsorted_data)
h = n//2

t0 = time.time()
sorted_data = insertion_sort(unsorted_data.copy(), n,0)
t1 = time.time()-t0
t0 = time.time()
sorted_data = shell_sort(unsorted_data.copy(), n, h,0)
t2 = time.time()-t0
# Calculate speedup percentage
# Using a max(t1, 1e-9) to avoid division by zero errors on tiny datasets
speedup = ((t1 - t2) / t1) * 100 if t1 > 0 else 0
print("-----------------------------")
print(f"insertion sort tme: {t1}")
print(f"shell sort time   : {t2}")
print(f"Shellsort speedup  : {speedup:.2f}%")
```

### **(4) Bubble Sort:**

Bubble Sort is a simple sorting algorithm that repeatedly steps through the list, compares adjacent elements, and swaps them if they are in the wrong order. The pass through the list is repeated until no swaps are needed, which indicates that the list is sorted. The algorithm gets its name from the way smaller elements 'bubble' to the top of the list.



#### **How it works**

- Start with the first element and compare it with the next element.
- If the current element is greater than the next element, swap them.
- Move to the next pair of elements and repeat steps 1-2.
- After the first pass, the largest element will be at the end of the array.
- Repeat the process for the remaining unsorted portion of the array until no swaps are made in a pass, indicating the array is sorted.

#### Example

Let's consider an array: `[5, 1, 4, 2, 8]`

- Pass 1:
  - (5, 1) -> (1, 5, 4, 2, 8)  (Swap 5 and 1)
  - (1, 5, 4, 2, 8) -> (1, 4, 5, 2, 8) (Swap 5 and 4)
  - (1, 4, 5, 2, 8) -> (1, 4, 2, 5, 8) (Swap 5 and 2)
  - (1, 4, 2, 5, 8) -> (1, 4, 2, 5, 8) (No swap for 5 and 8)
  - Array after Pass 1: `[1, 4, 2, 5, 8]`
- Pass 2:
  - (1, 4, 2, 5, 8) -> (1, 4, 2, 5, 8) (No swap for 1 and 4)
  - (1, 4, 2, 5, 8) -> (1, 2, 4, 5, 8) (Swap 4 and 2)
  - (1, 2, 4, 5, 8) -> (1, 2, 4, 5, 8) (No swap for 4 and 5)
  - Array after Pass 2: `[1, 2, 4, 5, 8]`
- Pass 3:
  - (1, 2, 4, 5, 8) -> (1, 2, 4, 5, 8) (No swap for 1 and 2)
  - (1, 2, 4, 5, 8) -> (1, 2, 4, 5, 8) (No swap for 2 and 4)
  - Array after Pass 3: `[1, 2, 4, 5, 8]`

No swaps occurred in Pass 3, so the array is sorted.

#### **Pseudocode**

Algorithm BubbleSort:

$$\begin{array}{ll}
\hline
\mathbf{Algorithm:} & \text{BubbleSort}(L,n) \\
\hline
 & \textbf{for } i \leftarrow 0 \text{ to } n-2 \textbf{ do} \\
 & \quad \textbf{for } j \leftarrow 0 \text{ to } n-i-2 \textbf{ do} \\
 & \quad \quad \textbf{if } L[j] > L[j+1] \textbf{ then} \\
 & \quad \quad \quad \text{swap}(L[j], L[j+1]) \quad \color{gray}{\text{// Swap adjacent elements}} \\
\hline
\end{array}$$

#### **Algorithm analysis:**

Bubble Sort is one of the simplest sorting algorithms, primarily used for educational purposes.
- Comparisons:
  - In Bubble Sort, the number of comparisons is fixed regardless of the initial order of the data (unless an optimized version with a swap flag is used). For an array of size $n$, the first pass performs $n-1$ comparisons, the second pass $n-2$, and so on.
  - Total comparisons for an array of size $n$:$$\sum_{i=1}^{n-1} (n-i) = (n-1) + (n-2) + \dots + 1 = \frac{n(n-1)}{2} \approx \frac{n^2}{2}$$
  - This holds true for the worst, average, and basic best cases.
- Exchanges:
  - Worst Case: If the array is in reverse order, every comparison results in a swap. The number of exchanges is equal to the number of comparisons: $\frac{n^2-n}{2} \approx \frac{n^2}{2}$.
   - Average Case: In a randomly ordered array, an exchange happens approximately half the time a comparison is made.$$\text{Average exchanges} \approx \frac{1}{2} \left( \frac{n(n-1)}{2} \right) = \frac{n^2 - n}{4} \approx \frac{n^2}{4}$$
   - Best Case: If the array is already sorted, 0 exchanges occur.
- Complexity Classes:
  - Worst-case Time $O(n^2)$: Occurs when the array is sorted in reverse order, requiring the maximum number of comparisons and swaps ($n^2/2$).
  - Average-case Time $\Theta(n^2)$: On average, the number of swaps is $n^2/4$, maintaining a quadratic growth rate.
  - Best-case Time $\Omega(n^2)$ or $\Omega(n)$: In the standard algorithm, it is $\Omega(n^2)$ because the loops run regardless of order. However, if the algorithm is optimized with a boolean flag to detect if a swap occurred, the best-case time becomes $\Omega(n)$ when the array is already sorted.
 - Space Complexity $O(1)$:It is an in-place sorting algorithm, requiring only a single additional memory space for the swap operation.

#### Python Code Example

```python
import random

def bubble_sort(L, n, logging=1):
    for i in range(n-1):
        # Last i elements are already in place
        for j in range(n - i - 1):
            if logging:
                print(f"i:{i}, j:{j} | L before: {L}")
            if L[j] > L[j + 1]:
                # Pythonic swap (Exchange)
                L[j], L[j + 1] = L[j + 1], L[j]
                if logging:
                    print(f"i:{i}, j:{j} | L after:  {L} (Swapped)")
            else:
                if logging:
                    print(f"i:{i}, j:{j} | L after:  {L} (No Swap)")

    print("\nBubble sorting is completed!")
    return L

# Example Usage:
unsorted_data = [64, 25, 12, 22, 11]
n = len(unsorted_data)

sorted_data = bubble_sort(unsorted_data.copy(), n, logging=1)

print("-----------------------------")
print(f"unsorted_data: {unsorted_data}")
print(f"Sorted array : {sorted_data}")
```

### **(5) Radix Sort**




Radix Sort is a non-comparative integer sorting algorithm that sorts data with integer keys by grouping the keys by individual digits that share the same significant position and value.

#### **How it works**

- Instead of comparing elements directly, Radix Sort distributes elements into buckets based on each digit's value.
- It typically starts from the least significant digit (LSD) and moves towards the most significant digit (MSD).
- A stable sorting algorithm (like Counting Sort) is used as a subroutine to sort the elements by each respective digit.
- After processing all digits, the array is completely sorted.

#### **Example:**

Assuming we have a list: [170, 45, 75, 90, 802, 24, 2, 66]

- First Pass (Sort by 1s place):
  - Numbers are sorted by their last digit: [170, 90, 802, 2, 24, 45, 75, 66]
- Second Pass (Sort by 10s place):
  - Numbers are sorted by their middle digit: [802, 2, 24, 45, 66, 170, 75, 90]
- Third Pass (Sort by 100s place):
  - Numbers are sorted by their first digit: [2, 24, 45, 66, 75, 90, 170, 802]
- The array is now fully sorted.

#### **Pseudocode:**

Algorithm RadixSort:

$$
\begin{array}{l}
   \hline  \mathbf{Algorithm:} \text{ RadixSort}(L, n) \\   \hline
       \quad m \leftarrow \text{maximum value in } L \\
       \quad \text{exp} \leftarrow 1 \\
       \quad \textbf{while } m / \text{exp} > 0 \textbf{ do:} \\
       \quad \quad \text{CountingSort}(L, n, \text{exp}) \\
       \quad \quad \text{exp} \leftarrow \text{exp} \times 10 \\
    \hline
\end{array}
$$

#### **Algorithm analysis:**

Radix Sort is highly efficient for large datasets of integers or strings where the maximum number of digits is relatively small.

- Comparisons:
  - Radix sort does not use comparisons. It relies on the distribution of elements.
- Complexity Classes:
  - Worst-case Time $$O(d(n + k))$$: Where $$d$$ is the number of digits, $$n$$ is the number of elements, and $$k$$ is the base of the number system (e.g., 10 for decimal).
  - Average-case Time $$\Theta(d(n + k))$$: Performance remains consistent regardless of the initial order.
  - Best-case Time $$\Omega(d(n + k))$$: Even if sorted, it must process every digit of every number.
- Space Complexity $$O(n + k)$$: It is not an in-place algorithm. The underlying Counting Sort requires additional arrays for counting and outputting elements.

#### **Python code**

```python
def counting_sort_for_radix(L, exp, logging=1):
    n = len(L)
    output = [0] * n
    count = [0] * 10

    # Store count of occurrences
    for i in range(n):
        index = L[i] // exp
        count[index % 10] += 1

    # Change count[i] so that it contains the actual position
    for i in range(1, 10):
        count[i] += count[i - 1]

    # Build the output array
    i = n - 1
    while i >= 0:
        index = L[i] // exp
        output[count[index % 10] - 1] = L[i]
        count[index % 10] -= 1
        i -= 1

    # Copy the output array to L
    for i in range(n):
        L[i] = output[i]

    if logging:
        print(f"List after sorting by digit place {exp}: {L}")

def radix_sort(L, n, logging=1):
    # Find the maximum number to know the number of digits
    max_val = max(L)
    exp = 1
    while max_val // exp > 0:
        counting_sort_for_radix(L, exp, logging)
        exp *= 10
    print("\nRadix sorting is completed!")
    return L

# Example Usage:
unsorted_data = [170, 45, 75, 90, 802, 24, 2, 66]
n = len(unsorted_data)
sorted_data = radix_sort(unsorted_data.copy(), n)
print("-----------------------------")
print(f"unsorted_data: {unsorted_data}")
print(f"Sorted array : {sorted_data}")
```

### **(6) Merge Sort**

Merge Sort is a highly efficient, stable, divide-and-conquer algorithm. It works by repeatedly dividing the unsorted list into sublists until each sublist contains one element, and then repeatedly merging these sublists to produce new sorted sublists.


#### **How it works**

- **Divide:** Split the unsorted list into two approximately equal halves.
- **Conquer:** Recursively sort both halves using Merge Sort. A list of 1 element is considered sorted.
- **Combine (Merge):** Merge the two sorted halves back together into a single sorted list by comparing the smallest unpicked elements of each half.

#### **Example:**

Assuming we have a list: [38, 27, 43, 3, 9, 82, 10]

- Split into [38, 27, 43, 3] and [9, 82, 10]
- Split further until single elements: [38], [27], [43], [3], [9], [82], [10]
- Merge [38] and [27] -> [27, 38]
- Merge [43] and [3] -> [3, 43]
- Merge [27, 38] and [3, 43] -> [3, 27, 38, 43]
- Merge [9] and [82] -> [9, 82]
- Merge [9, 82] and [10] -> [9, 10, 82]
- Final Merge: [3, 27, 38, 43] and [9, 10, 82] -> [3, 9, 10, 27, 38, 43, 82]

#### **Pseudocode:**

Algorithm MergeSort:

$$
\begin{array}{l}
   \hline  \mathbf{Algorithm:} \text{ MergeSort}(L, \text{left}, \text{right}) \\   \hline
       \quad \textbf{if } \text{left} < \text{right} \textbf{ then} \\
       \quad \quad \text{mid} \leftarrow \text{left} + (\text{right} - \text{left}) // 2 \\
       \quad \quad \text{MergeSort}(L, \text{left}, \text{mid}) \\
       \quad \quad \text{MergeSort}(L, \text{mid} + 1, \text{right}) \\
       \quad \quad \text{Merge}(L, \text{left}, \text{mid}, \text{right}) \\
    \hline
\end{array}
$$

#### **Algorithm analysis:**

Merge Sort guarantees a logarithmic time complexity, making it highly reliable for large datasets, though it requires extra memory.

- Comparisons:
  - The array is halved $$\log_2(n)$$ times. Merging two halves takes at most $$n$$ comparisons.
- Complexity Classes:
  - Worst-case Time $$O(n \log n)$$: The array is always divided and merged.
  - Average-case Time $$\Theta(n \log n)$$: Consistent performance regardless of initial order.
  - Best-case Time $$\Omega(n \log n)$$: Even if sorted, the algorithm still divides the array and compares elements during the merge phase.
- Space Complexity $$O(n)$$: It requires additional memory equal to the size of the input array to hold the split halves during the merge process.

#### **Python code**

```python
def merge_sort(L, logging=1):
    if len(L) > 1:
        mid = len(L) // 2
        left_half = L[:mid]
        right_half = L[mid:]

        if logging:
            print(f"Splitting: {L} -> {left_half} & {right_half}")

        merge_sort(left_half, logging)
        merge_sort(right_half, logging)

        i = j = k = 0

        # Merge the temp arrays back into L
        while i < len(left_half) and j < len(right_half):
            if left_half[i] < right_half[j]:
                L[k] = left_half[i]
                i += 1
            else:
                L[k] = right_half[j]
                j += 1
            k += 1

        # Checking if any element was left behind
        while i < len(left_half):
            L[k] = left_half[i]
            i += 1
            k += 1

        while j < len(right_half):
            L[k] = right_half[j]
            j += 1
            k += 1

        if logging:
            print(f"Merged   : {L}")

    return L

# Example Usage:
unsorted_data = [38, 27, 43, 3]
sorted_data = merge_sort(unsorted_data.copy())
print("-----------------------------")
print(f"unsorted_data: {unsorted_data}")
print(f"Sorted array : {sorted_data}")
```

### **(7) Quick Sort**


Quick Sort is a highly efficient, in-place, divide-and-conquer sorting algorithm. It is generally considered the fastest sorting algorithm in practice for general-purpose sorting.

#### **How it works**

- **Pivot Selection:** Choose an element from the array to act as the "pivot" (often the last element, first element, or a random element).
- **Partitioning:** Rearrange the array so that all elements smaller than the pivot are moved to its left, and all elements greater are moved to its right. The pivot is now in its final sorted position.
- **Recursion:** Recursively apply the same process to the sub-arrays on the left and right of the pivot.

#### **Example:**

Assuming we have a list: [10, 80, 30, 90, 40, 50, 70] and we choose the last element (70) as the pivot.

- Compare elements with 70. 10, 30, 40, 50 are smaller. 80, 90 are larger.
- Partition the array: [10, 30, 40, 50, **70**, 90, 80]
- 70 is now in its correct place.
- Recursively apply to the left side [10, 30, 40, 50] and right side [90, 80].
- Left side is already sorted. Right side uses 80 as pivot, swaps to [80, 90].
- Final Array: [10, 30, 40, 50, 70, 80, 90]

#### **Pseudocode:**

Algorithm QuickSort:

$$
\begin{array}{l}
   \hline  \mathbf{Algorithm:} \text{ QuickSort}(L, \text{low}, \text{high}) \\   \hline
       \quad \textbf{if } \text{low} < \text{high} \textbf{ then} \\
       \quad \quad \text{pi} \leftarrow \text{Partition}(L, \text{low}, \text{high}) \\
       \quad \quad \text{QuickSort}(L, \text{low}, \text{pi} - 1) \\
       \quad \quad \text{QuickSort}(L, \text{pi} + 1, \text{high}) \\
    \hline
\end{array}
$$

#### **Algorithm analysis:**

Quick Sort is incredibly fast on average, but its worst-case scenario can be problematic if the pivot is chosen poorly.

- Comparisons:
  - Heavily depends on the pivot. If the pivot splits the array roughly in half every time, it takes $$n \log n$$ comparisons.
- Complexity Classes:
  - Worst-case Time $$O(n^2)$$: Occurs when the array is already sorted (or reverse sorted) and the highest/lowest element is consistently chosen as the pivot.
  - Average-case Time $$\Theta(n \log n)$$: On average, the pivot divides the array into reasonably proportional sub-arrays.
  - Best-case Time $$\Omega(n \log n)$$: The pivot perfectly halves the array every time.
- Space Complexity $$O(\log n)$$: It is an in-place sort, but requires memory for the recursive call stack. In the worst case, this stack can grow to $$O(n)$$.


#### **Pyhton code**

```python
def partition(L, low, high, logging=1):
    pivot = L[high]
    i = low - 1  # Index of smaller element

    for j in range(low, high):
        if L[j] <= pivot:
            i = i + 1
            L[i], L[j] = L[j], L[i]

    # Swap the pivot element with the element at i+1
    L[i + 1], L[high] = L[high], L[i + 1]

    if logging:
        print(f"Pivot {pivot} placed at index {i+1}: {L[low:high+1]}")

    return i + 1

def quick_sort(L, low, high, logging=1):
    if low < high:
        # pi is partitioning index
        pi = partition(L, low, high, logging)

        # Separately sort elements before and after partition
        quick_sort(L, low, pi - 1, logging)
        quick_sort(L, pi + 1, high, logging)

    return L

# Example Usage:
unsorted_data = [10, 80, 30, 90, 40, 50, 70]
n = len(unsorted_data)
sorted_data = quick_sort(unsorted_data.copy(), 0, n - 1)
print("-----------------------------")
print(f"unsorted_data: {unsorted_data}")
print(f"Sorted array : {sorted_data}")
```

### **(8) Power Sort**


Power Sort is a modern, nearly-optimal merge sort variant. It identifies naturally occurring sorted "runs" in the data and merges them based on power-of-two intervals. It was adopted as the standard sorting algorithm in Python (replacing Timsort) starting in CPython 3.11.

#### **How it works**

- **Run Detection:** It scans the array to find existing sorted sub-arrays (runs). If a run is descending, it reverses it to make it ascending.
- **Node Powers:** It calculates a "power" for the boundary between two adjacent runs. This power represents the depth in a conceptual binary merge tree.
- **Merge Strategy:** It maintains a stack of pending runs. When a new run is found, it checks the power of the boundaries. If the power indicates that previous runs should be merged to maintain a balanced tree, it merges them before pushing the new run onto the stack.

#### **Example:**

Assuming an array with natural runs: [5, 6, 7, 1, 2, 3]

- Detects Run A: [5, 6, 7]
- Detects Run B: [1, 2, 3]
- Computes the merge power between Run A and Run B.
- Since they form adjacent nodes in the conceptual binary tree, Power Sort merges them.
- Merged result: [1, 2, 3, 5, 6, 7]

#### **Pseudocode:**

Algorithm PowerSort:

$$
\begin{array}{l}
   \hline  \mathbf{Algorithm:} \text{ PowerSort}(L, n) \\   \hline
       \quad \text{Stack} \leftarrow \text{Empty} \\
       \quad \textbf{while } \text{unprocessed elements remain} \textbf{ do:} \\
       \quad \quad R \leftarrow \text{Find next natural run in } L \\
       \quad \quad p \leftarrow \text{Compute power between top of Stack and } R \\
       \quad \quad \textbf{while } \text{Stack has runs and merge conditions are met based on } p \textbf{ do:} \\
       \quad \quad \quad \text{Merge top runs in Stack} \\
       \quad \quad \text{Push } R \text{ onto Stack} \\
       \quad \textbf{while } \text{Stack has more than one run} \textbf{ do:} \\
       \quad \quad \text{Merge remaining runs} \\
    \hline
\end{array}
$$

#### **Algorithm analysis:**

Power Sort excels in real-world data, which often contains partially sorted sequences.

- Comparisons:
  - Minimizes comparisons by leveraging existing order in the data.
- Complexity Classes:
  - Worst-case Time $$O(n \log n)$$: Similar to Merge Sort, but with less overhead in practice.
  - Average-case Time $$\Theta(n \log n)$$: Highly optimized merging strategy.
  - Best-case Time $$\Omega(n)$$: If the array is already sorted, it detects a single run and finishes immediately.
- Space Complexity $$O(n)$$: Requires memory for merging runs, similar to standard Merge Sort.

#### **Python code**

```python
def merge_runs(L, start, mid, end, logging=1):
    left = L[start:mid]
    right = L[mid:end]
    i = j = 0
    k = start

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            L[k] = left[i]
            i += 1
        else:
            L[k] = right[j]
            j += 1
        k += 1

    while i < len(left):
        L[k] = left[i]
        i += 1
        k += 1

    while j < len(right):
        L[k] = right[j]
        j += 1
        k += 1

    if logging:
        print(f"Merged runs into: {L[start:end]}")

def power_sort_simplified(L, n, logging=1):
    runs = []
    i = 0

    # 1. Detect natural runs
    while i < n:
        start = i
        i += 1
        while i < n and L[i] >= L[i-1]:
            i += 1
        runs.append((start, i))
        if logging:
            print(f"Detected run: {L[start:i]}")

    # 2. Merge runs (Simplified stack-based merge)
    while len(runs) > 1:
        # Pop two runs and merge them
        r1_start, r1_end = runs.pop(0)
        r2_start, r2_end = runs.pop(0)

        merge_runs(L, r1_start, r1_end, r2_end, logging)

        # Add the merged run back
        runs.insert(0, (r1_start, r2_end))

    print("\nPower sorting (simplified) is completed!")
    return L

# Example Usage:
unsorted_data = [5, 6, 7, 1, 2, 3, 9, 8]
n = len(unsorted_data)
sorted_data = power_sort_simplified(unsorted_data.copy(), n)
print("-----------------------------")
print(f"unsorted_data: {unsorted_data}")
print(f"Sorted array : {sorted_data}")
```

## **Sorting Algorithms Comparison: Time Performance**

To truly understand the impact of time complexity, we can benchmark all the sorting algorithms we've covered against the same dataset.

In the code below, we generate a random list of 1,000 integers. We then pass a copy of this list to each algorithm, measure the time it takes to complete the sort, and rank the results from fastest (best time) to slowest.

*(Note: We use a custom `HiddenPrints` context manager to suppress the step-by-step logging output from our previously defined functions, ensuring the benchmark runs cleanly without freezing the notebook).*



### **Python Code for Benchmarking**


```python
import random
import time
import sys
import os

# Context manager to suppress print statements from the sorting functions
class HiddenPrints:
    def __enter__(self):
        self._original_stdout = sys.stdout
        sys.stdout = open(os.devnull, 'w')
    def __exit__(self, exc_type, exc_val, exc_tb):
        sys.stdout.close()
        sys.stdout = self._original_stdout

# Generate a dataset large enough to show performance differences
n = 1000
unsorted_data = [random.randint(1, 5000) for _ in range(n)]

# Dictionary to store execution times
results = {}

def benchmark(name, func, *args):
    """Helper function to time the sorting algorithms."""
    with HiddenPrints():
        t0 = time.time()
        func(*args)
        t1 = time.time()
    results[name] = t1 - t0

# Run benchmarks (passing 0 for 'logging' where applicable)
benchmark("Selection Sort", selection_sort, unsorted_data.copy(), n, 0)
benchmark("Insertion Sort", insertion_sort, unsorted_data.copy(), n, 0)
benchmark("Shell Sort", shell_sort, unsorted_data.copy(), n, n//2, 0)
benchmark("Bubble Sort", bubble_sort, unsorted_data.copy(), n, 0)
benchmark("Radix Sort", radix_sort, unsorted_data.copy(), n, 0)
benchmark("Merge Sort", merge_sort, unsorted_data.copy(), 0)
benchmark("Quick Sort", quick_sort, unsorted_data.copy(), 0, n - 1, 0)
benchmark("Power Sort", power_sort_simplified, unsorted_data.copy(), n, 0)

# Sort results by execution time (fastest to slowest)
sorted_results = sorted(results.items(), key=lambda x: x[1])

# Display the results
print(f"--- Sorting Algorithm Performance Comparison (n={n}) ---")
print(f"{'Algorithm':<18} | {'Time (seconds)':<15}")
print("-" * 36)
for name, exec_time in sorted_results:
    print(f"{name:<18} | {exec_time:.6f}")
```

### **Takeaways from the Benchmark**

- **The $O(n \log n)$ Champions:** Quick Sort, Merge Sort, and Power Sort consistently group at the top. They handle 1,000 items in fractions of a millisecond.
- **The Non-Comparative Advantage:** Radix Sort often wins outright on integer datasets because it bypasses element-to-element comparisons entirely, running in $O(d(n+k))$ time.
- **The $O(n^2)$ Bottleneck:** Insertion, Selection, and Bubble Sort trail significantly behind. As $n$ increases to 10,000 or 100,000, these three will take exponentially longer (seconds or even minutes), while the top algorithms will still finish in milliseconds.
- **The Middle Ground:** Shell Sort bridges the gap. By pre-sorting with gaps, it drastically improves upon standard Insertion Sort without the complex recursive overhead of Quick or Merge Sort.

**Note about Power Sort:**

If we changed the benchmark to use a dataset that was 90% sorted with a few random elements at the end, the results would flip entirely.

In that scenario, Quick Sort might struggle, Radix Sort would take the exact same amount of time as before, but Power Sort would finish almost instantly (approaching $O(n)$ best-case time) because it would detect the massive existing run and simply merge the few remaining items. Power Sort is the "real-world" champion, even if it loses the "pure random data" synthetic benchmark!
